## Task 1: Conceptual Understanding

### Q1. Difference between "Love" and "love" in NLP?

Without lowercasing, NLP models treat them as **two different tokens**, 
increasing vocabulary size and reducing accuracy.

- "Love" → start of sentence or proper noun
- "love" → regular verb/noun

Lowercasing solves this → both become "love"

---

### Q2. What happens if stopwords are not removed?

Stopwords (is, the, and, a) add noise but carry little meaning.
Without removing them:
- Vocabulary size increases unnecessarily
- Important words lose importance
- Model performance drops

---

### Q3. Two scenarios where removing stopwords is harmful

**1. Sentiment Analysis:**
- "I do not like this" → removes "not" → "I like this" ❌
- Meaning is completely reversed

**2. Search Queries:**
- "What is the capital of India?" → "capital India"
- Question intent and structure is lost ❌

---

### Q4. Stemming vs Lemmatization

| Feature | Stemming | Lemmatization |
|---------|----------|---------------|
| Method | Rule-based | Dictionary-based |
| Speed | Fast | Slow |
| Accuracy | Lower | Higher |
| Output | May not be real word | Always real word |
| Example | "studies" → "studi" ❌ | "studies" → "study" |

### Combine code

```python
import re
import unicodedata

def preprocess_text(text):

    # Task 7: Error Handling — Start
    if not isinstance(text, str):
        return ""
    if not text.strip():
        return ""
    
    # Step 1: Remove URLs and emails
    text = re.sub(r'http\S+|www\S+', '', text)
    text = re.sub(r'\S+@\S+', '', text)
    
    # Step 2: Remove emojis
    text = ''.join(char for char in text if unicodedata.category(char) != 'So')
    
    # Step 3: Lowercase
    text = text.lower()
    
    # Step 4: Handle repeated characters
    text = re.sub(r'(.)\1{2,}', r'\1', text)
    
    # Step 5: Remove numbers
    text = re.sub(r'\d+', '', text)
    
    # Step 6: Remove punctuation
    text = re.sub(r'[^\w\s]', '', text)
    
    # Step 7: Tokenize
    tokens = text.split()
    
    # Step 8: Remove short tokens (keep "no", "not")
    meaningful = ["no", "not"]
    tokens = [word for word in tokens if word in meaningful or len(word) > 2]
    
    # Step 9: Remove extra spaces
    text = ' '.join(tokens)

    # Task 7: Error Handling — End
    if not tokens:
        return ""
    
    return tokens
```

```python
print(preprocess_text("Get 100% FREE access now!!!"))
print(preprocess_text("I absolutely looooved this product 😍😍"))
print(preprocess_text("Visit https://openai.com now!"))
print(preprocess_text("I am not happy with this"))
```

```python
sample_inputs = [
    "Get 100% FREE access now!!!",
    "I absolutely looooved this product 😍😍",
    "Worst service ever... 0/10",
    "Call me at 9876543210",
    "This is THE best course!!!",
    "Visit https://openai.com now!",
    "Nooooo this is baaad!!!",
    "OK OK OK I got it",
    "Win $$$ now!!! Limited offer!!!",
    "I am not happy with this"
]
```

## Task 4: Token Analytics

```python
for sentence in sample_inputs:
    tokens = preprocess_text(sentence)
    clean_sentence = ' '.join(tokens)

    total = len(tokens)
    unique = len(set(tokens))

    avg_length = round(sum(len(w) for w in tokens) / total,2)
    print(f"Original        : {sentence}")
    print(f"Tokens          : {tokens}")
    print(f"Total Tokens    : {total}")
    print(f"Unique Tokens   : {unique}")
    print(f"Avg Token Length: {avg_length}")
    print("-" * 50)
```

## Task 5: Frequency Analysis

```python
from collections import Counter

all_tokens = []

for sentence in sample_inputs:
    tokens = preprocess_text(sentence)
    all_tokens.extend(tokens)

print(all_tokens)
```

```python
word_counts = Counter(all_tokens)

# Top 10 most frequent
print("Top 10 Most Frequent:")
print(word_counts.most_common(10))

# Top 5 least frequent
print("Top 5 Least Frequent:")
print(word_counts.most_common()[:-6:-1])
```

##  Task 6: Build Full Pipeline

```python
def full_pipeline(text_list):
    
    all_tokens = []
    clean_sentences = []
    
    for sentence in text_list:
        tokens = preprocess_text(sentence)
        clean_sentence = ' '.join(tokens)
        
        all_tokens.extend(tokens)
        clean_sentences.append(clean_sentence)  
    
    return {
        "tokens": all_tokens,
        "clean_sentences": clean_sentences
    }
```

```python
result = full_pipeline(sample_inputs)

print("Tokens:")
print(result["tokens"])

print("\nClean Sentences:")
for sentence in result["clean_sentences"]:
    print(sentence)
```
